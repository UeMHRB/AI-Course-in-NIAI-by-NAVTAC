# Task 2: Store Numbers

## Description
This task demonstrates basic arithmetic operations in Python by adding two numbers together after getting them as an input by the user.

### What the code does:
- Get two numbers as inputs from the user 
- Performs addition operation on both numbers and stores them in another variable named sum
- Displays the result in an output string



```python
num1 = input("Enter the first number: ")
num2 = input("Enter the second number: ")

sum = float(num1) + float(num2)
print("The sum of", num1, "and", num2, "is:", sum)
```
